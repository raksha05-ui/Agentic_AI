```python
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
```

```python
c = canvas.Canvas("telecom_guide.pdf", pagesize=letter)
c.drawString(100, 750, "Hello, this is your Telecom Guide!")
c.drawString(100, 730, "More content can be added here...")
c.save()
print("PDF 'telecom_guide.pdf' created successfully!")
```

When using `reportlab`, to 'add' content to an existing PDF, you typically need to recreate the entire document, including all previous content and the new additions. The `canvas.save()` method will overwrite the file if it already exists. Below, I'll recreate the `telecom_guide.pdf` with expanded content.

```python
c = canvas.Canvas("telecom_guide.pdf", pagesize=letter)

c.drawString(100, 750, "Hello, this is your Telecom Guide!")
c.drawString(100, 730, "More content can be added here...")

c.drawString(100, 680, "--- Section 1: Understanding Telecom ---")
c.drawString(100, 660, "Telecommunications refers to the exchange of information over significant distances...")
c.drawString(100, 640, "It encompasses technologies like telephone, radio, television, and the internet.")

c.drawString(100, 590, "--- Section 2: Key Telecom Services ---")
c.drawString(100, 570, "1. Voice Calls: Traditional landlines and mobile voice services.")
c.drawString(100, 550, "2. Data Services: Broadband internet, mobile data, and enterprise networks.")
c.drawString(100, 530, "3. Messaging: SMS, MMS, and over-the-top (OTT) messaging apps.")

c.save()
print("PDF 'telecom_guide.pdf' recreated with additional content successfully!")
```

```python
from langchain_community.document_loaders import PyPDFLoader
pdf_path="telecom_guide.pdf"
loader = PyPDFLoader(pdf_path)

docs = loader.load()

print(f"Successfully loaded {len(docs)} pages from {pdf_path}.")
for i, doc in enumerate(docs):
    print(f"--- Page {i+1} ---")
    print(doc.page_content[:500])
    if i == 0 and len(docs) > 1:
        print("... (truncated)")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=60,chunk_overlap=0)
split_docs=splitter.split_documents(docs)
```

```python
!pip install langchain-huggingface langchain-chroma
!pip install opentelemetry-api==1.42.1 opentelemetry-sdk==1.42.1 --force-reinstall
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

db = Chroma.from_documents(split_docs, embeddings)
```

```python
data=db.as_retriever(search_type="similarity", search_kwargs={"k":2})
```

```python
query="Whats telecom"
response=data.invoke(query)
for i, doc in enumerate(response,1):
    print(f"--- Page {i+1} ---")
    print(doc.page_content[:500])
```
